```python
import pandas as pd
import numpy as np

data = pd.Series([1, np.nan, 'hello', None])

print(data.isnull())
print(data[data.notnull()])
```

```python
df = pd.DataFrame(np.arange(1, 21).reshape(5, 4),columns=['A', 'B', 'C', 'D'])

df.iloc[0, 1] = np.nan
df.iloc[1, 2] = np.nan
df.iloc[2, 0] = np.nan
df.iloc[4, 3] = np.nan

print(df)
print(df.isnull().sum())
```

```python
print(df.dropna())
print(df.dropna(axis='columns'))
```

```python
df = pd.DataFrame({'A': [1, np.nan, np.nan],'B': [np.nan, 2, np.nan],'C': [3, 4, np.nan]})

print(df.dropna(how='all'))
```

```python
print(df.dropna(thresh=3))
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'Day': ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'],'Temperature': [30, 32, np.nan, 35, np.nan, 34, 36]})

print(df)

df_mean = df.copy()
df_mean['Temperature'] = df_mean['Temperature'].fillna(df_mean['Temperature'].mean())
print(df_mean)

df_interpolate = df.copy()
df_interpolate['Temperature'] = df_interpolate['Temperature'].interpolate()
print(df_interpolate)
```

```python
import pandas as pd

df = pd.DataFrame({'department': ['CSE', 'CSE', 'AIML', 'AIML', 'ECE', 'ECE','CSE', 'CSE', 'AIML', 'AIML', 'ECE', 'ECE'],'gender': ['M', 'F', 'M', 'F', 'M', 'F','F', 'M', 'F', 'M', 'F', 'M'],'salary': [40000, 45000, 50000, 55000, 35000, 38000,42000, 43000, 52000, 54000, 36000, 39000]})

table = pd.pivot_table(df, values='salary', index='gender',columns='department', aggfunc='mean')

print(table)
```

```python
table = pd.pivot_table(df, values='salary', index='gender',columns='department', aggfunc='count')

print(table)
```

```python
table = pd.pivot_table(df, values='salary', index='gender',columns='department', aggfunc='mean',margins=True)

print(table)
```

```python
df['age'] = [22, 25, 28, 32, 35, 38, 24, 27, 31, 34, 40, 45]

df['age_category'] = pd.cut(df['age'], bins=[0, 25, 35, 100],labels=['Junior', 'Mid', 'Senior'])

table = pd.pivot_table(df, values='salary', index='age_category',columns='department', aggfunc='mean',observed=False)

print(table)
```

```python
table = pd.pivot_table(df, index='department',values=['salary', 'age'],aggfunc={'salary': 'sum', 'age': 'mean'})

print(table)
```
