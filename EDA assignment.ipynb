```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

X = np.random.rand(10, 2)

print(X)
print("Shape:", X.shape)
```

```python
plt.scatter(X[:, 0], X[:, 1])

plt.xlabel("X coordinate")
plt.ylabel("Y coordinate")
plt.title("10 Random Points")

plt.show()
```

```python
diff = X[:, np.newaxis, :] - X[np.newaxis, :, :]

print("Coordinate differences:")
print(diff)
print("Shape:", diff.shape)
```

```python
squared_diff = diff ** 2

print("Squared differences:")
print(squared_diff)
print("Shape:", squared_diff.shape)
```

```python
squared_diff = diff ** 2

print("Squared differences:")
print(squared_diff)

print("Shape:", squared_diff.shape)
```

```python
dist_sq = np.sum(squared_diff, axis=2)

print("Summed squared distances:")
print(dist_sq)

print("Shape:", dist_sq.shape)
```

```python
diagonal = np.diag(dist_sq)

print("Diagonal of distance matrix:")
print(diagonal)
```

```python
neighbor_order = np.argsort(dist_sq, axis=1)

print("Neighbor ordering:")
print(neighbor_order)
```

```python
print("First column:")
print(neighbor_order[:, 0])
```

```python
K = 2

nearest = np.argpartition(dist_sq, K, axis=1)[:, :K+1]

print("K+1 nearest points (including itself):")
print(nearest)
```

```python
neighbors = nearest[:, 1:]

print("Two nearest neighbors:")
print(neighbors)
```

```python
plt.figure(figsize=(7, 7))

# Plot the 10 points
plt.scatter(X[:, 0], X[:, 1], s=80)

# Label each point
for i in range(10):
    plt.text(X[i, 0], X[i, 1], str(i))

# Draw lines to the 2 nearest neighbors
for i in range(10):
    for j in neighbors[i]:
        plt.plot(
            [X[i, 0], X[j, 0]],
            [X[i, 1], X[j, 1]]
        )

plt.xlabel("X coordinate")
plt.ylabel("Y coordinate")
plt.title("2-Nearest Neighbor Network")

plt.show()
```

```python
def loop_knn(X, K=2):
    N = len(X)
    neighbors = []

    for i in range(N):
        distances = []

        for j in range(N):
            diff = X[i] - X[j]
            distance = np.sum(diff ** 2)
            distances.append(distance)

        order = np.argsort(distances)
        neighbors.append(order[1:K+1])

    return np.array(neighbors)
```

```python
loop_result = loop_knn(X, 2)

print("Loop-based nearest neighbors:")
print(loop_result)
```

```python
np.random.seed(42)

X100 = np.random.rand(100, 2)
X1000 = np.random.rand(1000, 2)
X5000 = np.random.rand(5000, 2)

print(X100.shape)
print(X1000.shape)
print(X5000.shape)
```

```python
def vectorized_knn(X, K=2):
    dist_sq = np.sum(
        (X[:, np.newaxis, :] - X[np.newaxis, :, :]) ** 2,
        axis=2
    )

    # Find K+1 smallest values, including the point itself
    candidates = np.argpartition(dist_sq, K, axis=1)[:, :K+1]

    # Sort only these candidates
    candidate_dist = np.take_along_axis(dist_sq, candidates, axis=1)
    order = np.argsort(candidate_dist, axis=1)
    candidates = np.take_along_axis(candidates, order, axis=1)

    # Remove the point itself
    return candidates[:, 1:K+1]
```

```python
print("Vectorized nearest neighbors:")
print(vectorized_knn(X100, 2))
```

```python
print("Vectorized method:")
%timeit vectorized_knn(X100, 2)

print("\nLoop-based method:")
%timeit loop_knn(X100, 2)
```

```python
print("Vectorized method:")
%timeit vectorized_knn(X1000, 2)

print("\nLoop-based method:")
%timeit loop_knn(X1000, 2)
```
