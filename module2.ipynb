```python
import pandas as pd

population = pd.Series(
    [20, 19, 13, 8, 15],
    index=["Mumbai", "Delhi", "Bengaluru", "Chennai", "Kolkata"]
)

print(population)

print("Population of Bengaluru:")
print(population["Bengaluru"])

print("Delhi to Chennai:")
print(population["Delhi":"Chennai"])
```

```python
import pandas as pd

products = {
    "Pen": 10,
    "Book": 50,
    "Bag": 500
}
prices = pd.Series(products)
print(prices)
print(prices.index)

print("Dictionary keys:")
print(products.keys())
print(prices.index.equals(pd.Index(products.keys())))
```

```python
import pandas as pd

area = pd.Series([307713, 191791, 130060, 114084, 342239],index=["Rajasthan", "Karnataka", "Tamil Nadu", "Kerala", "Maharashtra"])
population = pd.Series([80, 68, 77, 35, 126],index=["Rajasthan", "Karnataka", "Tamil Nadu", "Kerala", "Maharashtra"])
states = pd.DataFrame({"area": area,"population": population})
states["density"] = states["population"] / states["area"]
print(states)
```

```python
import pandas as pd

indA = pd.Index([2, 3, 5, 7, 11])
indB = pd.Index([1, 3, 5, 7, 9])

print("Intersection:")
print(indA.intersection(indB))

print("Union:")
print(indA.union(indB))

print("Symmetric Difference:")
print(indA.symmetric_difference(indB))
```
