```python
import numpy as np

data = np.array([
    ("Wings of Fire", "A.P.J. Abdul Kalam", 250.0),
    ("Harry Potter", "J.K. Rowling", 400.0),
    ("The Alchemist", "Paulo Coelho", 300.0),
    ("Atomic Habits", "James Clear", 500.0)
], dtype=[("title", "U30"), ("author", "U20"), ("price", "f4")])

print(data)
```

```python
print(data["title"])
```

```python
print(data[1])
```

```python
amount = 350

print(data["title"][data["price"] < amount])
```

```python
import numpy as np

dtype1 = {
    "names": ["title", "author", "price"],
    "formats": ["U30", "U20", "f4"]
}

dtype1 = np.dtype(dtype1)

print(dtype1)
```

```python
dtype2 = np.dtype([
    ("title", "U30"),
    ("author", "U20"),
    ("price", "f4")
])

print(dtype2)
```

```python
print(dtype1 == dtype2)
```

```python
import numpy as np

data_rec = data.view(np.recarray)

print(data_rec.price)
```

```python
import numpy as np

students = np.array([
    (1, "Rahul", 75.5),
    (2, "Anu", 89.0),
    (3, "Kiran", 68.5),
    (4, "Sneha", 95.0),
    (5, "Arjun", 82.0)
], dtype=[
    ("roll_no", "i4"),
    ("name", "U20"),
    ("marks", "f4")
])

students = np.sort(students, order="marks")[::-1]

print(students)
```
