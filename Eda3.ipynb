```python
import numpy as np

arr = np.arange(1, 21).reshape(4, 5)
print(arr)
view = arr[:3, :4]
view[0, 0] = 100
print(view)
print(arr)
copy_arr = arr[:3, :4].copy()
copy_arr[0, 0] = 200
print(copy_arr)
print(arr)
```

```python
import numpy as np
x = np.arange(1, 13)
print(x)
matrix = x.reshape(3, 4)
print(matrix)
col1 = x.reshape(12, 1)
print(col1)
col2 = x[:, np.newaxis]
print(col2)
print(np.array_equal(col1, col2))
```

```python
import numpy as np

np.random.seed(0)
a = np.random.randint(1, 10, (2, 3))
b = np.random.randint(1, 10, (2, 3))
print(a)
print(b)
v1 = np.vstack((a, b))
v2 = np.concatenate((a, b), axis=0)
print(v1)
print(v2)
print(np.array_equal(v1, v2))
h1 = np.hstack((a, b))
h2 = np.concatenate((a, b), axis=1)
print(h1)
print(h2)
print(np.array_equal(h1, h2))
```
